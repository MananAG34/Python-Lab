```python
#1.Write a Python function to find the maximum and minimum numbers from a sequence
# of numbers. (Note: Do not use built-in functions.)
def max(n):
    max = n[0]
    for i in (n):
        if(i > max):
           max = i
    return max
def min(n):
    min = n[0]
    for i in (n):
        if(i < min):
          min = i
    return min
n = [1,2,3,4,7,6,5,9,-1]
print(f"{max(n),min(n)}=(max,min)")

```

```python
# 2. Write a Python function that takes a positive integer and returns the sum of the cube of
# all the positive integers smaller than the specified number.
def sumcube(n):
    sum=0
    for i in range(1,n,):
        sum+=i**3
    return sum
n=int(input())
print(sumcube(n))




```

```python
# Fibonacci using Recursion

def fib(n):
    if n == 0:          # base case: fib(0) = 0
        return 0
    if n == 1:          # base case: fib(1) = 1
        return 1
    return fib(n-1) + fib(n-2)   # recursive call

n = int(input("Enter n: "))

# Print full series from 0 to n
for i in range(n + 1):
    print(fib(i), end=" ")
```

```python
#q.7 Keyword arguments: passed by name, order doesn't matter

def student_info(name, age, grade):
    # Displaying student details received as keyword arguments
    print(f"Name  : {name}")
    print(f"Age   : {age}")
    print(f"Grade : {grade}")

# Calling with keyword arguments — order can be anything
student_info(age=20, grade="A", name="Alice")
```

```python
# Default arguments: used when no value is passed for that parameter

def greet(name, message="Hello"):   # message has a default value
    # If message is not provided, "Hello" is used automatically
    print(f"{message}, {name}!")

greet("Bob")                        # uses default message
greet("Alice", "Good Morning")      # overrides default message
```

```python
#variable length arguments
# *args   → accepts any number of positional arguments (stored as tuple)
# **kwargs → accepts any number of keyword arguments (stored as dict)

def show_items(*args):
    # *args lets us pass any number of positional values
    print("Items passed:")
    for item in args:
        print(f"  - {item}")

def show_details(**kwargs):
    # **kwargs lets us pass any number of key=value pairs
    print("Details passed:")
    for key, value in kwargs.items():
        print(f"  {key} : {value}")

# Calling with variable number of arguments
show_items("apple", "banana", "cherry", "mango")

show_details(name="Alice", city="Delhi", age=21)
```

```python
#Q.8check if all values in a dictionary are the same
# Lambda extracts unique values using set(); if set has only 1 element → all same
all_same = lambda d: len(set(d.values())) == 1

# Test Case 1: all values are the same
d1 = {"a": 10, "b": 10, "c": 10}

# Test Case 2: values are different
d2 = {"a": 10, "b": 20, "c": 10}

# Test Case 3: single key-value pair (trivially same)
d3 = {"x": 99}

# Displaying results
print(f"d1 = {d1}")
print(f"All values same? → {all_same(d1)}")   # True

print(f"\nd2 = {d2}")
print(f"All values same? → {all_same(d2)}")   # False

print(f"\nd3 = {d3}")
print(f"All values same? → {all_same(d3)}")   # True
```

```python
#Q.9 Create two lists and combine them into a dictionary
# list1 → keys,  list2 → values

# Taking list inputs from user
n = int(input("Enter number of elements: "))

list1 = []   # will store keys
list2 = []   # will store values

print(f"\nEnter {n} keys (list1):")
for i in range(n):
    key = input(f"  Key {i+1}   : ").strip()
    list1.append(key)

print(f"\nEnter {n} values (list2):")
for i in range(n):
    val = input(f"  Value {i+1} : ").strip()
    list2.append(val)

# zip() pairs elements from both lists, dict() converts them to a dictionary
result = dict(zip(list1, list2))

# Display results
print(f"\nList 1 (keys)   : {list1}")
print(f"List 2 (values) : {list2}")
print(f"Generated Dict  : {result}")
```
