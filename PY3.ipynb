```python
# 1. Factorial of a number
n = int(input()) # asking user for an input
res = 1 # intialsing the result variable to 1 as if we put it to zero our answer would always be equal to zero
if (n<0): #factorial of negative numbers aren't defined.
  print("invalid response")
elif(n==0): # handling edge case factorial(0) =1
    print(1)
else:
    for i in range(n):
        res *= (n-i) # this is the main factorial calculation
print(res)


```

```python
# 2. Armstrong Number
n = int(input())
s = str(n) # converting the number to string so that we can get it's length
l = len(s)
res=0
temp = n
for i in range(0,len(s),1):
    digit = temp%10 # getting the digit
    temp//=10 # changing n from like 153 -> 15
    res+=digit**len(s)
if res==n:
   print("Amrstrong")
else:
    print("Not armstrong")





```

```python
# 4. Prime Number Check
def prime(n):
    for i in range(2,int(n**1/2) +1 ,1):
        if(n%i == 0):
           return True
n = int(input())
print("Not prime") if prime(n) and n>=2 else print("Prime")

```

```python
# 5. Palindrome Check

n = int(input())
temp = n
rev = 0
while temp > 0:
    digit = temp % 10
    rev = rev * 10 + digit
    temp //= 10

print("Palindrome" if rev == n else "Not Palindrome")


```

```python
# 6. Sum of Digits
def digit(n):
    temp = n
    digits = []
    while(temp>0):
          digits.append(temp%10)
          temp//=10
    return sum(digits)
n = int(input())
print(digit(n))



```

```python
#8. Convert all lower cases to upper case in a string.
def upper(s):

    result = ""
    for ch in s:
        if ch.islower():
         result += ch.upper()
        else:
              result += ch

    print(result)

upper(input())
#print(s.lower()) build in function
```

```python
#10. 123454321
#.   1234 * 4321
#.   123 * * 321
#.   12 * * * 21
#.   1 * * * * 1
n = 5  # total rows = 5, also the middle number in the first row

for i in range(n):  # i starts at 0 and goes to 4, each i = one row

    # ── LEFT SIDE: print numbers from 1 up to (n-i) ──
    # row 0: prints 1,2,3,4,5  (n-0 = 5)
    # row 1: prints 1,2,3,4    (n-1 = 4)
    # row 2: prints 1,2,3      (n-2 = 3)
    # as i increases, we print fewer numbers each row
    for j in range(1, n-i+1):  # +1 because range is exclusive at the end
        print(j, end="")       # end="" means no newline, stay on same line

    # ── MIDDLE: print stars with a space before each ──
    # row 0: 0 stars  (i=0, so loop doesn't run at all)
    # row 1: 1 star   → " *"
    # row 2: 2 stars  → " * *"
    # stars increase as i increases
    for j in range(i):         # runs i times (0 times on row 0, 1 time on row 1...)
        print(" *", end="")    # space before * to match the pattern spacing

    # ── RIGHT SIDE: print numbers counting DOWN back to 1 ──
    # special case for row 0: we start from n-1 (=4) not n (=5)
    # because 5 is already printed on the left side, we don't want "123455 4321"
    # for all other rows: start from n-i since that's where left side ended
    start = n-1 if i == 0 else n-i  # ternary: if first row use n-1, else use n-i

    for j in range(start, 0, -1):  # count down from start to 1 (0 is exclusive)
        print(j, end="")

    print()  # move to next line after each row is done
```
