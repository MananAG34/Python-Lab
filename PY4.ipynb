```python
# 4. WAP to enter a string and a substring. You have to print the number of times that the
# substring occurs in the given string. String traversal will take place from left to right, not
# from right to left.
# Sample Input
# ABCDCDC
# CDC
# Sample Output
# 2
string = input("Enter the string: ")
substring = input("Enter the substring: ")

count = 0
i = 0

while i <= len(string) - len(substring):
    if string[i : i + len(substring)] == substring:
        count += 1
        i += 1          # move by 1 to allow overlapping matches
    else:
        i += 1

print(count)

```

```python
# 5. Given a string containing both upper and lower case alphabets. Write a Python program
# to count the number of occurrences of each alphabet (case insensitive) and display the
# same.
# Sample Input
# ABaBCbGc
# Sample Output
# 2A
# 3B
# 2C
# 1G

string = input("Enter the string: ")

freq = {}

for char in string.upper():
    if char.isalpha():
        freq[char] = freq.get(char, 0) + 1

for key in sorted(freq):
    print(f"{freq[key]}{key}")
```

```python
#6.Program to count number of unique words in a given sentence using sets.

sentence = input("Enter a sentence: ")

words = sentence.lower().split()
unique_words = set(words)

print(f"Total words       : {len(words)}")
print(f"Unique words      : {len(unique_words)}")
print(f"Unique word list  : {unique_words}")
```

```python
# Q7: Create 2 sets of n fruits and perform set operations

n = int(input("Enter number of fruits (n): "))

# Taking n fruits for Set 1 (one fruit per input)
print(f"\nEnter {n} fruits for Set 1:")
s1 = {input(f"  Fruit {i+1}: ").lower() for i in range(n)}

# Taking n fruits for Set 2
print(f"\nEnter {n} fruits for Set 2:")
s2 = {input(f"  Fruit {i+1}: ").lower() for i in range(n)}

print(f"\ns1 = {s1}")
print(f"s2 = {s2}")

# a) Fruits in both s1 and s2 (Intersection)
common = s1 & s2
print(f"\na) Fruits in BOTH s1 and s2    : {common}")

# b) Fruits only in s1 but not in s2 (Difference)
only_s1 = s1 - s2
print(f"b) Fruits ONLY in s1 (not s2) : {only_s1}")

# c) Count of all unique fruits from s1 and s2 (Union)
all_fruits = s1 | s2
print(f"c) Count of ALL fruits (s1∪s2): {len(all_fruits)}")

```

```python
# 8. Take two sets and apply various set operations on them :
# S1 = {Red ,yellow, orange , blue }
# S2 = {violet, blue , purple}
S1 = {"Red", "yellow", "orange", "blue"}
S2 = {"violet", "blue", "purple"}

print(f"S1 = {S1}")
print(f"S2 = {S2}")

# 1. Union
print(f"\n1) Union (S1 | S2)            : {S1 | S2}")

# 2. Intersection
print(f"2) Intersection (S1 & S2)     : {S1 & S2}")

# 3. Difference S1 - S2
print(f"3) Difference (S1 - S2)       : {S1 - S2}")

# 4. Difference S2 - S1
print(f"4) Difference (S2 - S1)       : {S2 - S1}")
# 5. Symmetric Difference
print(f"5) Symmetric Diff (S1 ^ S2)   : {S1 ^ S2}")
```
