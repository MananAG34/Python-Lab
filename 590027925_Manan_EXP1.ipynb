1.Interactive Mode (The Shell)

The Interactive Mode is a live "chat" with Python.

    Symbol: Look for the >>> prompt.

    Behavior: Code runs instantly after you press Enter.

    Saving: No. Once you close the window, the code is gone.

    Best For: Quick math, testing one line of code, and debugging.

2.Script Mode (The Editor)

The Script Mode is a text editor for building actual programs.

    Access: File > New File.

    Behavior: Code runs only after you save as a .py file and press F5.

    Saving: Yes. Your work is permanent and reusable.

    Best For: Writing full projects, complex logic, and assignments.

6. Hypotenuse (Pythagoras Theorem)

```python
import math

a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = math.sqrt(a**2 + b**2)

print(f"The hypotenuse is: {c}")
```

8. Area of Triangle (Heron's Formula)

```python
import math

a = float(input("Side 1: "))
b = float(input("Side 2: "))
c = float(input("Side 3: "))

s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))

print(f"Area of the triangle: {area}")
```

9. Seconds to Hours, Minutes, Seconds

```python
total_sec = int(input("Enter seconds: "))

hours = total_sec // 3600
rem_sec = total_sec % 3600
minutes = rem_sec // 60
seconds = rem_sec % 60

print(f"{hours}h : {minutes}m : {seconds}s")
```

12. Bitwise Truth Table

```python
print("A | B | A&B | A|B | A^B")
print("-" * 22)
for a in [0, 1]:
    for b in [0, 1]:
        print(f"{a} | {b} |  {a&b}  |  {a|b}  |  {a^b}")
```

13. Bitwise Shift Operators

```python
num = int(input("Enter a number: "))
shift = int(input("Enter shift value: "))

print(f"Left Shift ({num} << {shift}): {num << shift}")
print(f"Right Shift ({num} >> {shift}): {num >> shift}")
```
