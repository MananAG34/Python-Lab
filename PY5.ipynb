```python
#1.Scan n values in range 0-3 and print the number of times each value has occurred


n = int(input("How many values to enter: "))

count = {0: 0, 1: 0, 2: 0, 3: 0}

for i in range(n):
    val = int(input(f"Enter value {i+1} (0-3): "))
    if val in count:
        count[val] += 1
    else:
        print(f"  {val} is out of range, skipping.")

print("\nOccurrence Count:")
for key, value in count.items():
    print(f"  {key} → {value} time(s)")
```

```python
#2.Create a tuple to store n numeric values and find average of all values.

n = int(input("Enter number of values: "))

values = tuple(float(input(f"Enter value {i+1}: ")) for i in range(n))

average = sum(values) / len(values)

print(f"Tuple   : {values}")
print(f"Average : {average:.2f}")
```

```python
# 3.WAP to find the runner-up score from a list of N student scores
# Runner-up = second highest unique score

n = int(input("Enter N: "))
scores = list(map(int, input("Enter scores separated by spaces: ").split()))

# Validate that exactly N scores were entered
if len(scores) != n:
    print(f"Error: Expected {n} scores but got {len(scores)}.")

# Need at least 2 unique scores to find a runner-up
elif len(set(scores)) < 2:
    print("Error: Need at least 2 different scores to find a runner-up.")

else:
    # Remove duplicates with set(), then sort in descending order
    # Example: [2,3,6,6,5] → {2,3,5,6} → [6,5,3,2]
    unique_scores = sorted(set(scores), reverse=True)

    max_score  = unique_scores[0]   # highest score
    runner_up  = unique_scores[1]   # second highest score

    print(f"\nScores entered : {scores}")
    print(f"Unique scores  : {sorted(set(scores), reverse=True)}")
    print(f"Max Score      : {max_score}")
    print(f"Runner-Up      : {runner_up}")
```

```python
# Q4: Person-City dictionary with operations

n = int(input("Enter number of persons: "))
persons = {}

for i in range(n):
    name = input(f"Enter name {i+1}: ").strip()
    city = input(f"Enter city of {name}: ").strip()
    persons[name] = city

# a) All names
print("\na) All Names:")
for name in persons.keys():
    print(f"   {name}")

# b) All cities
print("\nb) All Cities:")
for city in persons.values():
    print(f"   {city}")

# c) Name and city of each person
print("\nc) Name → City:")
for name, city in persons.items():
    print(f"   {name} → {city}")

# d) Count of persons per city
city_count = {}
for city in persons.values():
    city_count[city] = city_count.get(city, 0) + 1

print("\nd) Persons per City:")
for city, count in city_count.items():
    print(f"   {city} → {count} person(s)")
```

```python
# Q5: Movie details stored in a nested dictionary

n = int(input("Enter number of movies: "))
movies = {}

for i in range(n):
    print(f"\n--- Movie {i+1} ---")
    name       = input("  Name            : ").strip()
    year       = int(input("  Release Year    : "))
    director   = input("  Director        : ").strip()
    cost       = float(input("  Production Cost : "))
    collection = float(input("  Earnings        : "))

    movies[name] = {
        "year"      : year,
        "director"  : director,
        "cost"      : cost,
        "collection": collection,
        "profit"    : collection - cost   # profit = earning - cost
    }

# a) All movie details
print("\na) ALL MOVIES:")
for name, info in movies.items():
    print(f"  {name} | {info['year']} | Dir: {info['director']} | "
          f"Cost: {info['cost']} | Earned: {info['collection']} | Profit: {info['profit']}")

# b) Movies released before 2015
print("\nb) Released Before 2015:")
for name, info in movies.items():
    if info['year'] < 2015:
        print(f"  {name} ({info['year']})")

# c) Profitable movies
print("\nc) Profitable Movies:")
for name, info in movies.items():
    if info['profit'] > 0:
        print(f"  {name} → Profit: {info['profit']}")

# d) Movies by a particular director
director_search = input("\nd) Enter director name to search: ").strip().lower()
print(f"   Movies by '{director_search}':")
for name, info in movies.items():
    if info['director'].lower() == director_search:
        print(f"   {name} ({info['year']})")
```

```python
# Q6: Contact book using dictionary

contacts = {}  # key = name, value = phone number

def add_contact():
    name  = input("  Name  : ").strip()
    phone = input("  Phone : ").strip()
    if name in contacts:
        print(f"  '{name}' already exists. Use Update.")
    else:
        contacts[name] = phone
        print(f"  ✔ '{name}' added!")

def search_contact():
    name = input("  Enter name to search: ").strip()
    if name in contacts:
        print(f"  ✔ {name} → {contacts[name]}")
    else:
        print(f"  ✘ '{name}' not found.")

def update_contact():
    name = input("  Enter name to update: ").strip()
    if name in contacts:
        contacts[name] = input(f"  New phone for {name}: ").strip()
        print(f"  ✔ '{name}' updated!")
    else:
        print(f"  ✘ '{name}' not found.")

def delete_contact():
    name = input("  Enter name to delete: ").strip()
    if name in contacts:
        del contacts[name]
        print(f"  ✔ '{name}' deleted.")
    else:
        print(f"  ✘ '{name}' not found.")

def view_all():
    if not contacts:
        print("  Contact book is empty.")
    else:
        for name, phone in contacts.items():
            print(f"  {name:<20} {phone}")

while True:
    print("\n1.Add  2.Search  3.Update  4.Delete  5.View All  6.Exit")
    choice = input("Choose: ").strip()

    if   choice == '1': add_contact()
    elif choice == '2': search_contact()
    elif choice == '3': update_contact()
    elif choice == '4': delete_contact()
    elif choice == '5': view_all()
    elif choice == '6': print("Goodbye!"); break
    else: print("  Invalid choice.")
```

```python
# Q7: Todo list manager using a list

tasks = []  # list to store all tasks

def add_task():
    task = input("  Enter task: ").strip()
    if task:
        tasks.append(task)
        print(f"  ✔ Added: '{task}'")
    else:
        print("  Task cannot be empty.")

def view_tasks():
    if not tasks:
        print("  No tasks! List is empty.")
    else:
        print(f"  {len(tasks)} task(s):")
        for idx, task in enumerate(tasks, start=1):
            print(f"    {idx}. {task}")

def remove_task():
    view_tasks()
    if tasks:
        try:
            num = int(input("  Task number to remove: "))
            if 1 <= num <= len(tasks):
                removed = tasks.pop(num - 1)   # pop by index (0-based)
                print(f"  ✔ Removed: '{removed}'")
            else:
                print("  Invalid number.")
        except ValueError:
            print("  Enter a valid number.")

while True:
    print("\n1.Add Task  2.View Tasks  3.Remove Task  4.Exit")
    choice = input("Choose: ").strip()

    if   choice == '1': add_task()
    elif choice == '2': view_tasks()
    elif choice == '3': remove_task()
    elif choice == '4': print("Stay productive!"); break
    else: print("  Invalid choice.")
```
